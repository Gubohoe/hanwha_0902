```python
a = 33
b = 200

# 200 > 33 => True
if b > a:
  # 실행함
  print("b가 더 크다")

# 람다
if b > a:print("람다 >>> b가 더 크다")
```

```python
c = 33
d = 200

# 200 < 33 => False
if d < c:
  # 실행하지 않음
  print("d가 더 작다")

# 람다
print("람다 >>> d가 더 작다") if d < c else print("람다 >>> c가 더 작다")
```

```python
num1 = 50
num2 = 100

if num1 > num2:
  print("num1이 num2보다 큽니다.")
elif num1 == num2:
  print("num1과 num2는 같습니다.")
else:
  print("num1이 num2보다 작습니다")
```

```python
score = 75

if score >= 90:
  print("A")
elif score >= 80:
  print("B")
elif score >= 70:
  print("C")
elif score >= 60:
  print("D")
else:
  print("F")
```

match ... case 구문

```python
day = 4

match day:
  case 1:
    print("월요일")
  case 2:
    print("화요일")
  case 3:
    print("수요일")
  case 4:
    print("목요일")
  case 5:
    print("금요일")
  case 6:
    print("토요일")
  case 7:
    print("일요일")
  case _:
    print("그 밖의 날")
```

```python
d = 4

match d:
  case 1 | 2 | 3 | 4 | 5:
    print("주중입니다.")
  case 6 | 7:
    print("주말입니다,")
  case _:
    print("그 밖의 날")
```

While문

```python
i = 0

while i < 6:
  print(i)
  i += 1
```

```python
i = 0
while i < 6:
  if i == 3:
    break
  i += 1
  print(i)
```

```python
i = 0
while i < 6:
  i += 1

  if i == 3:
    continue
  print(i)
```

```python
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i는 6보다 크거나 같다.")
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
for x in "banana":
  print(x)
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    break
  print(x)
```

```python
# 2차원은 [행x, 열y]이라고 생각하자
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```
