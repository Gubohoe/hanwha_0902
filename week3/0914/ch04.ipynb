```python
from langchain_openai import ChatOpenAI
```

```python
# llm 객체 생성
llm = ChatOpenAI(
    temperature=0.1,
    model="gpt-4o-mini",
)
```

```python
# 질문
question = "미국의 수도는 어디인가요?"
```

```python
# 응답
response = llm.invoke(question)
```

```python
print(f"[답변]: {response}")
```

### logprob 로그 프로버블리티 활성화
- 모델의 답변 확신도(Confidence level) :0에 가까울수록 정답에 가까운 확신을 가진다.

```python
# 객체 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens=2048,
    model="gpt-4o-mini",
).bind(logprobs=True)
```

```python
question2 = "인도의 수도는 어디인가요?"

response = llm_with_logprob.invoke(question2)
```

```python
response
```

```python
response.response_metadata
```

```python
response.response_metadata["logprobs"]
```

### Stream 실시간 출력 방식

```python
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")
```

```python
for token in answer:
    print(token.content, end="", flush=True)
```

```python
answer2 = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")

final_answer = ""
for token in answer2:
    final_answer += token.content
```
