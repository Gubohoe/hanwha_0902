```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_classic.output_parsers import DatetimeOutputParser

from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

from enum import Enum
from langchain_classic.output_parsers.enum import EnumOutputParser
```

```python
output_parser = DatetimeOutputParser()
output_parser.format = "%Y-%m-%d"
```

```python
print(output_parser.get_format_instructions())
```

```python
template = """Answer the users question:\n\n

#Format Instructions: \n{format_instructions}\n\n
#Question: \n{question}\n\n
#Answer:"""

prompt = PromptTemplate.from_template(
    template,
    partial_variables={
        "format_instructions": output_parser.get_format_instructions()
    },
)

prompt
```

```python
template2 = """사용자의 질문에 답변해 주세요:\n\n

#형식 지정 지침: \n{format_instructions}\n\n
#질문: \n{question}\n\n
#답변:"""

prompt2 = PromptTemplate.from_template(
    template2,
    partial_variables={
        "format_instructions": output_parser.get_format_instructions()
    },
)

prompt2
```

```python
chain = prompt | ChatOpenAI() | output_parser
output = chain.invoke({"question": "Google이 창업한 연도"})
```

```python
print(output)
print(output.strftime("%Y-%m-%d"))
```

```python
class Colors(Enum):
    RED = "빨간색"
    GREEN = "초록색"
    BLUE = "파란색"
```

```python
Colors.RED
```

```python
parser = EnumOutputParser(enum=Colors)
parser.get_format_instructions()
```

```python
prompt = PromptTemplate.from_template(
    """다음의 물체는 어떤 색깔인가요?
    
    Object: {object}
    
    Instructions: {instructions}"""
).partial(instructions=parser.get_format_instructions())

chain = prompt | ChatOpenAI() | parser
```

```python
response = chain.invoke({"object": "하늘"})
print(response)
```
