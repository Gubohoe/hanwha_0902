```python
# 클래스 정의
class Person:
  # 클래스 초기화
  def __init__(self, name, age):
    self.name = name
    self.age = age

  # 메서드
  def greet(self):
    print("Hello, my name is " + self.name)

  # 출력메서드
  def display_info(self):
    print(f"이름:{self.name}, 나이:{self.age}")

# Obj 생성
p1 = Person("홍길동",16)
p2 = Person("벤자민",22)
p3 = Person("세종대왕",30)

print(p2.name)
print(p2.age)
p2.greet()
p2.display_info()
```
