```python
# 함수 선언(덧셈)
def cel(f):
  return f+100

# 함수 실행+출력
print(cel(77))
```

```python
# 사칙연산
def f1(su):
  return su+100
def f2(su):
  return su-100
def f3(su):
  return su*100
def f4(su):
  return su/100

print(f1(5))
print(f2(5))
print(f3(5))
print(f4(5))
```
